## Oppgave 1 - Medisinsk testing

En ny test for en sjelden sykdom er utviklet. Vi får følgende opplysninger:

__Begivenheter:__
* La $S$ betegne begivenheten med alle personer i befolkningen som har sykdommen.
* La $T$ betegne begivenheten med personer i befolkningen som vil teste positivt.

__Informasjon:__
* Sykdommen rammer $1\%$ av befolkningen.
* Testen gir positivt utslag for $95\%$ av alle personer som har sykdommen (sann positiv rate).
* Testen gir positivt utslag for $2\%$ av alle personer som ikke har sykdommen (falsk positiv rate).

**Spørsmål:** Hvis en person tar testen og får et positivt resultat, hva er sannsynligheten for at vedkommende faktisk har sykdommen?

**Bruk informasjonen over til først å angi med desimaltall:**
* $P(S) = ....$ (Andelen syke i befolkningen)
* $P(S^c) = ....$ (Andelen friske i befolkningen)
* $P(T|S) = ....$ (Sann positiv rate)
* $P(T|S^c) = ....$ (Falsk positiv rate)

**Du skal:** 
1. Lage et sannsynlighetstre som er tilpasset opplysningene i oppgaven. 
2. Beregne $P(S|T)$ - dvs sannsynligheten for at en person har sykdommen gitt at vedkommende tester positivt.

__Merk:__<br>
Løsningen av 2. krever bruk av Bayes' regel, så du bør først beregne $P(T)$. 


---
## *ANSWER:*

Bayes teorem er gitt som:

$$
P(S|T) = \frac{P(T|S) P(S)}{P(T)}
$$

da kan man først multiplisere andelen syke i befolknignen med andelen som får en sann positiv

$$
P(S) * P(T|S) = 0.01 * 0.95 = 0.0095
$$

For å få sannsynligheten for at personen faktisk er syk, trenger vi også de falske positive

$$
P(S^c) * P(T|S^c) = 0.99 * 0.02 = 0.0198
$$

Deretter addere disse for å få P(T)

$$
P(T) = 0.0095 + 0.0198 = 0.0293
$$

Nå setter vi inn i Bayes teorem og får

$$
P(S|T) = \frac{0.0095}{0.0293} = 0.3242
$$

Vi ender opp med en sannsynlighet på 32,4%

## Oppgave 2 - Terningspill med 3 rettferdige terninger
(Se tips til løsning nederst i oppgaven.)

### 2a) 
Vi kaster 3 rettferdige terninger og studerer følgende begivenheter:
- **Begivenhet $A$**: *Første terning viser et partall* (dvs. 2, 4 eller 6).  
- **Begivenhet $B$**: *Summen av de tre terningene er et oddetall*.  

**Oppgaver:**

1. Beskriv utfallsrommet $S$ for kast av 3 terninger (med antall utfall og hvordan de kan representeres).  
2. Bestem antall utfall i $A$ og antall utfall i $B$.  
3. Regn ut $P(A)$ og $P(B)$.  
4. Regn ut $P(A \cap B)$.  
5. Sjekk om $A$ og $B$ er uavhengige ved å undersøke om  
   $$ P(A \cap B) \stackrel{?}{=} P(A)\,P(B).$$

---

### 2b)
- **Begivenhet $C$**: *Summen av de tre terningene er lik 10*.  
- **Begivenhet $D$**: *Alle terningene viser forskjellige verdier*.  

**Oppgaver:**

1. Beskriv hvilke utfall som tilhører $C$ (alle kombinasjoner av tre terninger som summerer til 10).  
2. Beskriv hvilke utfall som tilhører $D$ (kombinasjoner der minst to av de tre terningene er like).  
3. Regn ut $P(C)$ og $P(D)$.  
4. Finn $P(C \cap D)$, altså sannsynligheten for at summen er 10 og minst to terninger viser samme verdi.  
5. Bestem den betingede sannsynligheten $P(D|C)$.
6. Undersøk uavhengighet ved å se om  
   $$P(C \cap D) \stackrel{?}{=} P(C) \, P(D).$$
7. Kan spørsmålet om uavhengighet også forklares fra den betingede sannsynligheten $P(D|C)$?

---

### Tips til løsning
- Et utfall av tre terningskast kan beskrives med et tall-trippel $(x_1,x_2,x_3)$, der koordinatverdiene $x_1, x_2, x_3\in \{1,2,3,4,5,6\}$.  
- For å finne sannsynligheter, bruk  
  $$
  P(\text{begivenhet}) = \frac{\text{antall utfall i begivenheten}}{\text{antall utfall i utfallsrommet}}.
  $$
- For å avgjøre om to begivenheter $A$ og $B$ er uavhengige, sjekk om  
  $$ P(A \cap B) = P(A)P(B).$$
  Hvis likheten ikke holder, er begivenhetene avhengige.

---

## ANSWER:

#### 2 a:

1. Hvis hver terning kan ha 6 utfall og vi har tre terninger har vi $|S| = 6^3 = 216$ muligheter.
2. Todelt oppgave:
    - I tilfelle A vil $\frac{1}{3}$ av tilfelle gi et partall, resten kan være hvilke som helst og dermed får vi;  $|S| = 3 * 6 * 6 = 108$ mulig løsninger.
    - For tilfelle B må enten 1 eller 3 av terningene være oddetall. Derfor skal vi se hvor mange muligheter de to tilfelle gir:
        - `3 Oddetall`: Hver terning vil gi oddetall halvparten av gangene. Antall muligheter $ 3*3*3 = 27$
        - `1 Oddetall`: Vi velger hvilken terning som skal være oddetall (3x3 valg), en oddetallsverdi for den (3x3 valg), og partallsverdier for de to andre terningene (3×3 valg). $3^4$ = 81
3. $ P(A) = \frac {108}{216} = 0.5 $, $ P(B) = \frac {108}{216} = 0.5$
4. Hvis første terning er et partall, må summen av de to neste tallene være oddetall. Dermed kan man først få hvilket som helst alternativ, men 3 på siste terning for å sørge for oddetall som sum. $ |A \cap B| = 3 * 6 * 3 = 54 $ som gir en sannsynlig $ P(A \cap B) = \frac {54}{216} = 0.25$
5. $P(A \cap B) = P(A) * P(B) = 0.5 * 0.5 = 0.25$, og dermed er de linært uavhengige. Det er altså ingen korrelasjon mellom de to hendelsene

#### 2 b:
1. Gjennom systematisk telling får jeg 27 muligheter. 
2. To av terningene kan være hvilken som helst, dermed 6x6 muligheter, men den siste har bare en mulighet. $|D| = 6 * 6 = 36$
3. $ P(C) = \frac {27}{216} = 0.125 $ og for to like: $ P(D) = \frac {36}{216} = 0.167 $ eller for alle ulike: $ P(D) = \frac {6*5*4}{216} = 0.556 $
4. Igjen, ved å telle, får vi 9 slike utfall. Dette gir: $ P(A \cap B) = \frac {9}{216} = 0.0417 $
5. Her brukes man Bayes Teorem og får $ P(D|C) = \frac {P(C \cap D)}{P(C)} = \frac{\frac{9}{216}}{\frac{27}{216}} = \frac{1}{3} = 0.33$
6. For å sjekke om de er uavhengige tar vi: $P(C)P(D) = \frac{27}{216} * \frac{36}{216} = \frac {1}{48} $
