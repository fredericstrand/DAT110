## 2 Oppsummering av data
Vi skal se på ulike nyttige oversiktsmetoder for å sammenfatte data både grafisk og numerisk. Vi legger vekt på viktigheten av å forstå data gjennom ulike statistiske verktøy (numeriske) og grafikk, som støtte for innsikt og påfølgende dataanalyse.

### [Grafiske teknikker](https://en.wikipedia.org/wiki/Plot_(graphics))

Grafiske teknikker er essensielle for å visualisere datafordelinger og identifisere mønstre, trender og uteliggere. Noen vanlige typer plott inkluderer:

- [**Spredningsplott**](https://en.wikipedia.org/wiki/Scatter_plot): Brukes til å vise verdier for vanligvis to variabler for et datasett. De hjelper med å identifisere forhold mellom variabler.
- [**Histogrammer**](https://en.wikipedia.org/wiki/Histogram): Viser fordelingen av et datasett ved å gruppere data i intervaller.
- [**Box plott**](https://en.wikipedia.org/wiki/Box_plot): Gir et grafisk sammendrag av sentraltendensen, variabiliteten og skjevheten til et datasett ved å vise 
    * [kvartiler](https://en.wikipedia.org/wiki/Quartile) og 
    * [uteliggere](https://en.wikipedia.org/wiki/Outlier).
- ([**Stem-and-Leaf Plots**](https://en.wikipedia.org/wiki/Stemplot): Ligner på histogrammer, men beholder de opprinnelige dataverdiene mens de viser fordelingen. Dekkes ikke i DAT110.)

----

### [Oppsummerende statistikk](https://en.wikipedia.org/wiki/Summary_statistics)

Numeriske oppsummeringer inkluderer mål som gjennomsnitt, median, modus, varians og standardavvik. Disse målene hjelper til med å forstå sentraltendensen og variabiliteten til dataene:

- [**Gjennomsnitt**](https://en.wikipedia.org/wiki/Arithmetic_mean): Gjennomsnittet av alle datapunktene.
- [**Median**](https://en.wikipedia.org/wiki/Median): Den midterste verdien når datapunktene er ordnet i rekkefølge.
- [**Modus**](https://en.wikipedia.org/wiki/Mode_(statistics)): Den hyppigst forekommende verdien i et datasett.
- [**Varians**](https://en.wikipedia.org/wiki/Variance): Måler hvor langt hvert tall i settet er fra gjennomsnittet.
- [**Standardavvik**](https://en.wikipedia.org/wiki/Standard_deviation): Kvadratroten av variansen, som gir et mål på spredningen av et sett med verdier.
- [__Kvartiler__](https://en.wikipedia.org/wiki/Quartile): De $3$ tallene $q_1$, $q_2$ og $q_3$ deler dataverdiene inn i fire intervaller som inneholder omtrent $25\%$ av dataene hver.
- [__Outliere__](https://en.wikipedia.org/wiki/Outlier): En outlier er et datapunkt som skiller seg betydelig fra andre observasjoner i et datasett.

### [Typer fordelinger](https://en.wikipedia.org/wiki/List_of_probability_distributions) (kapittel 4)

Å forstå forskjellige typer fordelinger er avgjørende for å analysere data. Noen viktige statistiske fordelinger inkluderer:

- [**Normalfordeling**](https://en.wikipedia.org/wiki/Normal_distribution): Ofte referert til som klokkekurven, er den symmetrisk rundt gjennomsnittet og karakterisert ved sitt gjennomsnitt ($\mu$) og standardavvik ($\sigma$). Den er mye brukt fordi mange variabler er tilnærmet normalfordelt.
- [**Uniform fordeling**](https://en.wikipedia.org/wiki/Uniform_distribution_(continuous)): Alle utfall er like sannsynlige; hver variabel har samme sannsynlighet.
- [**Binomialfordeling**](https://en.wikipedia.org/wiki/Binomial_distribution): Representerer antall suksesser i et fast antall uavhengige [Bernoulli-forsøk](https://en.wikipedia.org/wiki/Bernoulli_distribution#Related_distributions) med en konstant sannsynlighet for suksess.
- [**Poisson-fordeling**](https://en.wikipedia.org/wiki/Poisson_distribution): Modellerer antall hendelser innenfor et gitt tidsintervall (eller romlig område).

### [Histogrammer](https://en.wikipedia.org/wiki/Histogram)

<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/5/53/Cumulative_vs_normal_histogram.svg/350px-Cumulative_vs_normal_histogram.svg.png" width="600" height="300"/>

Histogrammer er effektive for å visualisere fordelingen av kvantitative numeriske data. For å konstruere et histogram er det første trinnet å "dele" hele verdiområdet inn i en serie med del-intervaller, – og deretter telle hvor mange verdier som faller inn i hvert del-intervall.

Et kumulativt histogram (som vist ovenfor) er en avbildning som teller det kumulative antallet observasjoner i alle intervallene opp til det spesifiserte intervallet. Det vil si at i et kumulativt histogram ($M$) er høyden på intervallet/søylen $M_i$ lik summen av intervall/søylehøydene $m_j, j\leq i$ i det ordinære histogrammet ($m$) som følger:

$$M_i=\sum_{j=1}^im_j.$$

__Form__<br>
Histogrammer lar oss se mønstre som skjevhet (asymmetri), modalitet (antall topper) og kurtose (hale). Å velge passende intervallbredder er avgjørende for å representere den underliggende fordelingen nøyaktig uten å overglatte eller underglatte.

Følgende histogrammer har en, to eller tre fremtredende topper. Slike fordelinger kalles henholdsvis unimodal, bimodal og multimodal. Enhver fordeling med mer enn 2 fremtredende topper kalles multimodal.

### Varians og standardavvik (2.1.4)

For et utvalg $x_1,..., x_n$ kan vi utlede [mål for sentraltendens](https://en.wikipedia.org/wiki/Central_tendency#Measures) ved følgende alternativer:

1. [Gjennomsnitt](https://en.wikipedia.org/wiki/Arithmetic_mean) ($\bar{x}$):
   $$
   \bar{x} = \frac{\sum x_i}{n}
   $$
2. [Median](https://en.wikipedia.org/wiki/Median): Midtpunktet når dataene er sortert.
3. [Modus](https://en.wikipedia.org/wiki/Mode_(statistics)): Den hyppigste verdien.

----

Vi kan også utlede forskjellige [mål for statistisk spredning](https://en.wikipedia.org/wiki/Statistical_dispersion), dvs.

1. [Utvalgsvarians](https://en.wikipedia.org/wiki/Variance#Sample_variance) ($s^2$):
   $$
   s^2 = \frac{\sum (x_i - \bar{x})^2}{n-1}
   $$
   * Utvalgsvariansen er per definisjon den gjennomsnittlige kvadrerte avstanden fra gjennomsnittet. Delingen med $n-1$ i stedet for $n$ vil bli forklart i kapittel 7 (se også [forklaringen på denne lenken](https://en.wikipedia.org/wiki/Variance#Biased_sample_variance)).

2. [Standardavvik](https://en.wikipedia.org/wiki/Standard_deviation) ($s$):
   $$
   s = \sqrt{s^2}
   $$
   * Standardavviket er per definisjon kvadratroten av utvalgsvariansen.

Standardavviket er nyttig når man vurderer hvor langt dataene er fordelt fra gjennomsnittet. Det representerer det typiske avviket til observasjoner fra gjennomsnittet.

Vanligvis (men ikke alltid) vil omtrent 70% av dataene være innenfor ett standardavvik fra gjennomsnittet, og omtrent 95% vil være innenfor to standardavvik.

3. [Kvartilene](https://en.wikipedia.org/wiki/Quartile) deler antall datapunkter i fire deler, eller kvartiler, av mer eller mindre lik størrelse. De tre kvartilene, som resulterer i fire datainndelinger, er følgende:

   * Den første kvartilen ($q_1$) er definert som den minste verdien der de laveste 25% av data-verdiene er under dette punktet.
   * Den andre kvartilen ($q_2$) er [medianverdien](https://en.wikipedia.org/wiki/Median) til et datasett (dermed ligger 50% av data-verdiene under dette punktet).
   * Den tredje kvartilen ($q_3$) er den minste verdien slik at 75% av data-verdiene er under dette punktet. Den er kjent som den øvre kvartilen.

[Det interkvartile området (IQR)](https://en.wikipedia.org/wiki/Interquartile_range) er definert som differansen:

   * IQR = $q_3 - q_1$.


----

```julia
# Vi kan beregne følgende mål for lokaasjon og spredning:
k   = 4; #
x   = iris[:,k];
vnk = vnames[k];

m₁ = mean(x)     # Sample mean
m₂ = median(x)   # Sample median
m₃ = mode(x)     # Sample mode
println("For $vnk har vi: mean = $(round(m₁,digits = 3)), median = $m₂ og mode = $m₃.")
println()

q₁ = quantile(x, 0.25) # First quartile
q₃ = quantile(x, 0.75) # Third quartile
iqr = q₃ - q₁    # Interquartile range
println("Første kvartil er $q₁, tredje kvartil er $q₃ og interquartil-bredden (-range) er $(round(iqr,digits = 3)).")
println()

s² = var(x)      # Sample variance
s = std(x)       # Sample standard deviation
println("Sampel-variansen er $(round(s²,digits = 3)) og sampel standard avviket er $(round(s,digits = 3)).")
```

### [Box (-og-whisker) Plott](https://en.wikipedia.org/wiki/Box_plot) og ekstremverdier

<img src="BoxPlotDef.png" width="500" height="400"/>

Boxplott gir et femtalls-sammendrag (minimum, første kvartil, median, tredje kvartil, maksimum) som hjelper til med å forstå spredningen og skjevheten i data. De er ofte nyttige for å identifisere uteliggere og sammenligne fordelinger på tvers av forskjellige grupper.

Et populært valg for grensene til [hårlinjene](https://en.wikipedia.org/wiki/Box_plot#Whiskers) er basert på 1,5 IQR-verdien. Fra over den øvre kvartilen (Q3) måles en avstand på 1,5 ganger IQR ut, og en _hårlinje tegnes opp til det største observerte datapunktet fra datasettet som faller innenfor_ denne avstanden. På samme måte måles en avstand på 1,5 ganger IQR ut under den nedre kvartilen (Q1), og en _hårlinje tegnes ned til det laveste observerte datapunktet fra datasettet som faller innenfor_ denne avstanden.

Fordi hårlinjene må ende på verdien av et observert datapunkt, kan hårlinjelengdene se ulike ut, selv om 1,5 IQR er den samme for begge sider. Alle andre _observerte datapunkter utenfor grensen til hårlinjene_ plottes som _outliere_.

----

#### Kontingenstabeller (2.2.1 - 2.2.2)

```julia
#using Pkg
#Pkg.add("FreqTables")

using FreqTables, StatsBase
# Make a contigency table
CT = freqtable(df,:homeownership, :application_type)
```

```julia

names(CT) # Get the names of the variables in the contingency table
parent(CT) # Get the parent of the contingency table, i.e. the data frame
permutedims(names(CT)[2]) # Permute the application type names to a row vector of strings
```

```julia
using DataFrames, StatsPlots
using FreqTables, StatsBase

# Make a contigency table
CT = freqtable(df,:homeownership, :application_type)

NM1 = names(CT)[1] # Get the names of the Home ownership variable in the contingency table.
NM2 = permutedims(names(CT)[2]) # Permute the names of the Application type variable in the contingency table into a row vector of strings. 

# Stacked bar plot
groupedbar(NM1, parent(CT), bar_position = :stack, bar_width=0.6, label = NM2, 
           xlabel="Home Ownership", ylabel="Count", title="Home Ownership grouped by Application Type")    
```

```julia
# Side-by-side version corresponding to the stacked bar plot
groupedbar(NM1, parent(CT), bar_position = :dodge, bar_width=0.6, label = NM2, 
           xlabel="Home Ownership", ylabel="Count", title="Home Ownership grouped by Application Type")    
```

```julia
# The marginal row proportions of the contigency table:
P₂CT = CT./sum(CT, dims = 2)
groupedbar(NM1,parent(P₂CT), bar_position = :stack, bar_width=0.6, label = NM2, xlabel="Home Ownership", 
           ylabel="Count", title="Home Ownership grouped by Application Type")    
```

### Case-studie: Malariavaksine
Vi ser på en studie på en ny malariavaksine kalt [_PfSPZ_](https://en.wikipedia.org/wiki/PfSPZ_Vaccine). I studien ble frivillige pasienter randomisert inn i en av to eksperimentgrupper:

1. 14 pasienter fikk en eksperimentell vaksine, og
2.  6 pasienter fikk en placebovaksine.

Nitten uker senere ble alle 20 pasientene eksponert for en medikamentfølsom malariavirusstamme (motivasjonen for å bruke en medikamentfølsom virusstamme her er av etiske hensyn, slik at eventuel smitte (infeksjon) kan behandles effektivt). Resultatene er oppsummert i følgende tabell:

| behandling | smittet  | ikke smittet | Totalt|
|-----------|-----------|--------------|-------|
| vaksine   | 5         | 9            | 14    |
| placebo   | 6         | 0            | 6     |
| Totalt    | 11        | 9            | 20    |

De observerte smitteandelene ($\frac{5}{14}=35.7\%$ for behandlingsgruppen versus $\frac{6}{6} = 100\%$ for kontrollgruppen) antyder at _vaksinen kan være effektiv_.

__Merk__<br>
Vi kan imidlertid ikke være sikre på om den observerte forskjellen representerer vaksinens effekt eller bare er tilfeldig. Generelt er det litt fluktuasjon i utvalgsdata, og vi forventer ikke at utvalgsproporsjonene skal være nøyaktig like, selv om sannheten var at smitteratene var uavhengige av å få vaksinen. I tillegg, med så små utvalg, er det kanskje vanlig å observere så store forskjeller når vi tilfeldig deler en gruppe på grunn av tilfeldigheter alene!

Det er tydelig at de observerte resultatene i datautvalget kanskje ikke perfekt gjenspeiler de sanne forholdene mellom variabler siden det er tilfeldig støy. Mens den observerte forskjellen i smitterater er stor, er utvalgsstørrelsen for studien liten, noe som gjør det uklart om denne observerte forskjellen representerer effekten av vaksinen eller om den rett og slett skyldes tilfeldigheter.

La oss merke disse to konkurrerende påstandene, $H_0$ og $H_A$, som uttales som "h-null" og "h-a":

* $H_0$: (Uavhengighetsmodell). Variablene `behandling` og `utfall` er _uavhengige_ - de har ingen sammenheng. (Den observerte forskjellen mellom andelen pasienter som ble smittet i de to gruppene, $64.3\%$, må ha vært tilfeldig.)
* $H_1$: (Alternativ modell.) Variablene _er ikke uavhengige_ (forskjellen i smitteandeler på $64.3\%$ var ikke tilfeldig, og vaksinen bidrar til å senke andelen smittede).

----

#### "Hands on" - Simulering av studien
Under antagelsen om at $H_0$ (uavhengighetsmodellen, dvs ingen sammenheng mellom `behandling` og `utfall`) er gyldig, kan vi simulere den naturlige variasjonen ved å tilfeldig fordele $11$ smittede og $9$ usmittede personer i to grupper på henholdsvis $14$ (vaksinerte) og $6$ (placebo behandlede) pasienter.

Fra et __stort antall simuleringer__ kan vi beregne andelen ganger som __differansen mellom andelen smittede pasienter i de to gruppene__ er minst like stor som det observerte $64.3\%$ eller større:


```julia
# Simulering av tilfeldig fordeling av 20 pasienter, 
# hvorav 11 er infiserte og 9 er ikke-infiserte, 
# inndelt i 2 grupper med størrelser på 14 og på 6:

using Random

N = 100;  # Totalt antall simuleringer
n = 20;   # Totalt antall pasienter

n1 = 14;  # Antall pasienter i gruppe 1 (vaksinegruppa)
n2 = 6;   # Antall pasienter i gruppe 2 (placebogruppa)

nnot = 9; # Antall usmittede pasienter
ninf = 11;# Antall smittede pasienter

observed = 6/n2 - 5/n1 # Den observerte differansen i andelen smittede mellom de to gruppene.

Patients = [zeros(nnot); ones(ninf)] # usmittede pasienter representeres med 9 nuller, smittede pasienter representeres med 11 enere.

res = zeros(N)

# Under H₀ antar vi at det ikke er noen sammenheng mellom vaksine og smitte, da kan vi N ganger
for k in 1:N
    SP = shuffle(Patients) # Stokker tilfeldig om på 0-er og 1-ere i Patients-vektoren.
    iv = sum(SP[1:n1])     # Summerer antall smittede blant de 14 første (vaksinegruppa).
    ip = sum(SP[n1+1:end]) # Summerer antall smittede blant de 6 siste (placebogruppa).
    res[k] = (ip/n2) - (iv/n1) # Andelen smittede i placebo-gruppa minus andelen smittede i vaksine-gruppa.
end

# Teller antall ganger differansen i smitte-andeler mellom de to gruppene er større enn eller lik 0.64:
test = sum(abs.(res) .≥ observed)/N

println("$(100*test)% av de simulerte differansene i andeler er større enn 0.64 som vi observerte i eksperimentet.")    

StackedDotPlot(res)

m = mean(res) # Gjennomsnittsverdien i simuleringene
# Legger til vertikal linje for gjennomsnitts-verdien (m) av simuleringene:
vline!([m], label="x̄ = $(round(m, digits=3)) (gjennomsnitt i simuleringene)", color=:red, linewidth=2)
```

__Konklusjon:__<br>
Det ser ut til at en forskjell på $64.3\%$ eller mer bare skjer i omtrent $2\%$ av simuleringene. En så lav sannsynlighet indikerer en sjelden hendelse, og med et $5\%$ ($\alpha = 0.05$) signifikansnivå forkaster vi derfor _nullhypotesen_ (__uavhengighetsmodellen__ $H_0$) til fordel for den alternative hypotesen ($H_1$) for å konkludere med at det er _signifikant bevis_ for at vaksinen gir en viss beskyttelse mot malariasmitte i denne kliniske settingen.

## Statistiske verktøy for dataanalyse (mer om dette senere)

Statistiske verktøy går utover enkel beskrivende statistikk og inkluderer slutningsstatistikk, som lar oss lage spådommer eller slutninger om en populasjon basert på et utvalg:

- **[Hypotesetesting](https://en.wikipedia.org/wiki/Statistical_hypothesis_test)**: Brukes for å avgjøre om det er nok bevis til å forkaste en nullhypotese til fordel for en alternativ hypotese.
- **[Konfidensintervaller](https://en.wikipedia.org/wiki/Confidence_interval)**: Gir et område der vi kan forvente at populasjonsparameteren ligger innenfor med et visst konfidensnivå.
- **[Regresjonsanalyse](https://en.wikipedia.org/wiki/Regression_analysis)**: Brukes til å modellere forhold mellom variabler og lage spådommer.
