Statistikk er basert på sannsynlighet, og selv om sannsynlighet ikke er nødvendig for de anvendte teknikkene i denne boken, vil det hjelpe deg med å få en dypere forståelse av metodene og legge et bedre grunnlag for fremtidig læring/kurs.

#### Eksempel

__Terningkast__<br>
En "terning" er en kube med seks sider nummerert 1, 2, 3, 4, 5 og 6.<br>
1. Hva er sjansen for å få 1 når du kaster en terning?
    * Svar: $\frac{1}{6}.$
2. Hva er sjansen for å få 1 eller 2 i neste kast?
    * Svar: $\frac{2}{6}=\frac{1}{3}.$
3. Hva er sjansen for å få enten 1, 2, 3, 4, 5 eller 6 i neste kast?
    * Svar: $\frac{6}{6}=1.$
4. Hva er sjansen for å ikke kaste en 2?
    * Svar: $1 - \frac{1}{6} = \frac{5}{6}.$

#### Grunnleggende sannsynlighet
Begrepet [sannsynlighet](https://en.wikipedia.org/wiki/Probability) er sjansen for at noe skal skje, kvantifisert som __et tall mellom 0 og 1__, hvor høyere verdier indikerer større sannsynlighet for forekomst. Men __hvordan__ beskriver vi sannsynligheter __formelt__?

Den vanlige måten er å vurdere et [sannsynlighetsrom](https://en.wikipedia.org/wiki/Probability_space), som matematisk består av tre elementer:
1. Et utfallsrom $Ω$ – en mengde hvis elementer tilsvarer alle mulige utfall av et bestemt eksperiment.
2. En samling $\Sigma = \{A_i\subseteq Ω\ | i\in Γ \}$ av målbare _begivenheter_, der hver begivenhet $A_i$ er en delmengde av utfallsrommet $\Omega$ ($\Sigma$ tilfredsstiller egenskapene til en [$\sigma$-algebra](https://en.wikipedia.org/wiki/%CE%A3-algebra#Definition)).
3. Et [sannsynlighetsmål](https://en.wikipedia.org/wiki/Probability_measure#Definition) $P:\Sigma\rightarrow [0,1]$, også betegnet her som _sannsynlighetsfunksjon_ – som angir sjansen $P(A_i)$ for at hver mulig begivenhet $A_i$ inntreffer.

---

I tillegg må 1.-3. tilfredsstille [__Sannsynlighetsaksiomene__](https://en.wikipedia.org/wiki/Probability_axioms):
* __A1__: $P(A_i)\geq 0$ for alle begivenheter $A_i\in \Sigma$.
* __A2__: $P(\Omega)=1$.<br>
For enhver endelig eller [tellbar](https://en.wikipedia.org/wiki/Countable) sekvens $A_1,\ A_2,\ ... $ av [disjunkte (ikke-overlappende) mengder](https://en.wikipedia.org/wiki/Disjoint_sets)<br>
* __A3__: $\mathbb{P}\left( \bigcup_{i = 1}^{\infty} A_i\right)=\sum_{i = 1}^{\infty}P(A_i)$.

__Merknad 1:__<br>
Hvis $A\subseteq B$, så er $P(A)\leq P(B)$, hvis $A\cap B = \emptyset$, så er $P(A\cup B)= P(A)+P(B)$, og for den tomme begivenheten $\emptyset$ har vi $P(\emptyset)=0$.

__Merknad 2:__ Ikke forveksle sannsynlighetsfunksjonen $\mathbb{P}$ med en sannsynlighetsmassefunksjon, som vi definerer nedenfor.

[__En tilfeldig (stokastisk) prosess__](https://en.wikipedia.org/wiki/Stochastic_process)<br>
... er en situasjon der vi __vet hvilke utfall__ som __kan skje__, men vi __vet ikke hvilket bestemt utfall__ som __vil skje__.

[Eksempler på tilfeldige prosesser](https://en.wikipedia.org/wiki/Stochastic_process#Examples) er:

* Aksjemarkedet (om det går opp eller ned)
* osv.

#### Eksempel
__Kasting av to terninger__
Betrakt det tilfeldige eksperimentet der to uavhengige, rettferdige, seks-sidede terninger kastes, og vi ønsker å finne sannsynligheten for at summen av utfallene av terningene er jevn (et partall). Her kan utfallsrommet representeres som
$$Ω = \{(i,j)\ |i, j ∈ {1,..., 6} \} = \{1,..., 6\}\times \{1,..., 6\},$$
dvs. det kartesiske produktet av mengden av enkeltkast-utfall med seg selv. Det vil si at elementer i utfallsrommet inneholder alle tupler av formen $(i, j)$ der $i, j ∈ {1,..., 6}$, og er listet i følgende tabell:

| Terning 1 \ Terning 2 | 1      | 2      | 3      | 4      | 5      | 6      |
|-----------------|--------|--------|--------|--------|--------|--------|
| **1**           | (1,1)  | (1,2)  | (1,3)  | (1,4)  | (1,5)  | (1,6)  |
| **2**           | (2,1)  | (2,2)  | (2,3)  | (2,4)  | (2,5)  | (2,6)  |
| **3**           | (3,1)  | (3,2)  | (3,3)  | (3,4)  | (3,5)  | (3,6)  |
| **4**           | (4,1)  | (4,2)  | (4,3)  | (4,4)  | (4,5)  | (4,6)  |
| **5**           | (5,1)  | (5,2)  | (5,3)  | (5,4)  | (5,5)  | (5,6)  |
| **6**           | (6,1)  | (6,2)  | (6,3)  | (6,4)  | (6,5)  | (6,6)  |

La oss si at vi er interessert i sannsynligheten for begivenheten:

$$A =\{(i,j)\ |\ i+j \text{ er jevn}\}.$$

I dette tilfeldige eksperimentet, siden terningene ikke har noen iboende skjevhet, er det fornuftig å anta en symmetrisk sannsynlighetsfunksjon. Det vil si at for enhver $B ⊂ Ω$,

$$P(B)=\frac{|B|}{|Ω|},$$

der $|·|$ teller antall utfall (elementer) i mengden. Det kalles symmetrisk fordi hvert utfall i $Ω$ har samme sannsynlighet. Den nøyaktige sannsynligheten er åpenbart

$$P(A)=\frac{18}{36}=0.5.$$

#### Monte Carlo-simulering med Julia
Nedenfor bruker vi Julia for å utforske dette videre via både en eksakt beregning og ved en såkalt [Monte Carlo-simulering](https://en.wikipedia.org/wiki/Monte_Carlo_method):
* Den eksakte beregningen gir den nøyaktige andelen av jevne summer.
* En Monte Carlo-simulering gjentar eksperimentet mange ganger og estimerer $\mathbb{P}(A)$ basert på andelen av ganger et utfall i begivenheten $A$ forekom.


```julia
# SwJ - Listing 2.1: Jevn sum av to terninger
# ---------------------------------------
n = 10^5;    # antall forsøk
faces = 1:6; # sidene på terningene

p = sum([iseven(i+j) for i in faces, j in faces])/length(faces)^2     # ekte brøkdel av jevne summer

@time p̂ₙ  = sum([iseven(rand(faces) + rand(faces)) for i in 1:n]) / n  # Monte Carlo-estimat av p


println("Eksakt løsning p = $p \nMonte Carlo-estimat p̂ₙ = $p̂ₙ")
```

```julia
# SwJ - Alternativ 2.1: Jevn sum av to terninger
# -------------------------------------------
using StatsBase
n, faces = 10^5, 1:6    # antall forsøk og sider på terningene

p = mean([(i+j)%2==0 for i in faces, j in faces])                # ekte brøkdel av jevne summer

@time p̂ₙ = mean(iseven.(rand(faces,n) + rand(faces,n)))          # Monte Carlo-estimat av p


println("Eksakt løsning p = $p \nMonte Carlo-estimat p̂ₙ = $p̂ₙ")
```

__Svar__: $\frac{|(1,1)|}{|\Omega|}=\frac{1}{36}.$

### [__Borel's lov om store tall__](https://en.wikipedia.org/wiki/Law_of_large_numbers#Borel's_law_of_large_numbers)

Etter hvert som flere ($n$) observasjoner av en tilfeldig prosess samles inn, vil andelen $\hat{p}_n = \frac{N_n(A)}{n}$ av forekomster i en bestemt begivenhet $A$ i løpet av $n$ forsøk, konvergere mot den sanne sannsynligheten $p=\mathbb{P}(A)$ for denne begivenheten, dvs.
$$\lim_{n\rightarrow \infty} \hat{p}_n = p.$$

Monte Carlo-simuleringen ovenfor gir en empirisk demonstrasjon av Borel's lov.

### __Sannsynlighet – "hverdagsdefinisjonen"__ 
Vi bruker sannsynlighet til å bygge verktøy for å beskrive og forstå tilsynelatende tilfeldighet. Vi formulerer ofte sannsynlighet i form av en __tilfeldig prosess__ som gir et __utfall__, som for eksempel:
* Kaste en terning: $\ \ \rightarrow 1,\ 2,\ 3,\ 4,\ 5,\ 6.$
* Kaste en mynt: $\ \ \ \ \ \ \rightarrow H\ \text{eller}\ T.$

Borels lov om store tall gir __den rigide formuleringen__ av vår intuitive definisjon av __sannsynligheten__ for en begivenhet $A$:<br>

"Sannsynligheten $P(A)$ er andelen av ganger som $A$ ville forekomme hvis vi observerte utfallet av et uendelig antall forsøk fra en tilfeldig prosess."

Ved å definere sannsynlighet som en andel, tar den alltid verdier mellom 0 og 1 (inkludert). Sannsynligheter angis ofte også som en prosentandel mellom $0\%$ og $100\%$.

#### Simulering
Sannsynlighet kan illustreres ved å kaste en terning mange ganger. Som ovenfor lar vi $\hat{p}_n$ være andelen av utfall (kast) som _er lik_ $1$ etter å ha gjennomført de første $n$ kastene.

Etter hvert som antall kast øker, vil $\hat{p}_n$ konvergere mot sannsynligheten $p = P(1)=\frac{1}{6}$. Følgende Julia-kode illustrerer denne konvergensen for $n = 100.000$ terningkast. 

Tendensen til at $\hat{p}_n$ stabiliserer seg rundt $p$ er i samsvar med Borels lov om store tall som er nevnt ovenfor.


```julia
n = 100_000;    # antall forsøk
faces = 1:6;    # sidene på terningene
p = 1/6         # sannsynlighet for å få en 1-er

p̂ₙ = cumsum(rand(faces,n).==1)./(1:n) # andel av jevne summer
#p̂ₙ = cumsum([rand(faces)==1 for i in 1:n])./(1:n) # andel av jevne summer (alternativ kode)

using Plots
plot(log10.(1:n), p̂ₙ, label="p̂ₙ", xlabel="log10(n - antall forsøk)", ylabel="p̂ₙ", title="Konvergens av p̂ₙ mot p", legend=:topleft)
plot!(log10.(1:n), p*ones(n), label="p = 1/6", linestyle=:dash)
```

#### Disjunkte eller gjensidig utelukkende begivenheter
To begivenheter kalles __disjunkte__ eller __gjensidig utelukkende__ hvis de ikke begge kan skje i samme forsøk. For eksempel, hvis vi kaster en terning, er begivenhetene
* "å kaste en 1" og "å kaste en 2" disjunkte siden begge ikke kan forekomme i ett kast.
* "å kaste en 1" og "å kaste et oddetall" ikke disjunkte siden begge forekommer hvis utfallet av kastet er en 1-er.

Begrepene __disjunkt__ og __gjensidig utelukkende__ er ekvivalente og kan brukes om hverandre.



---

Å beregne sannsynligheten for disjunkte utfall er enkelt. Når man kaster en terning, er utfallene 1 og 2 disjunkte, og vi beregner sannsynligheten for at et av disse utfallene vil forekomme ved å legge sammen deres separate sannsynligheter:
$$P(1 \text{ eller } 2) = P(1) + P(2) = 1/6 + 1/6 = 1/3.$$
Det samme gjelder for
$$P(1 \text{ eller } 2 \text{ eller } 3 \text{ eller } 4 \text{ eller } 5 \text{ eller } 6) = P(1) + P(2) + P(3) + P(4) + P(5) + P(6)$$
$$1/6+1/6+1/6+1/6+1/6+1/6=1.$$


__Addisjonsregelen for disjunkte begivenheter__<br>
    Hvis $A_1$ og $A_2$ representerer to disjunkte begivenheter, så er sannsynligheten for at en av dem inntreffer gitt ved
    $$P(A_1 \cup A_2)= P(A_1)+P(A_2).$$
    Hvis det er mange disjunkte utfall $A_1, ..., A_k$, så er sannsynligheten for at en av disse utfallene vil inntreffe
    $$P(A_1 \cup A_2 \cup \text{ ... }\cup A_k) = P(A_1)+P(A_2)+ ... + P(A_k).$$

#### Sannsynligheter når begivenheter ikke er disjunkte
Betrakt en kortstokk som vist i figur 3.3 i OIS,

<img src="CardDeck3_3.png"/>

og begivenhetene
* $A$ = "mengden av alle ruter-kort".
* $B$ = "mengden av alle bildekort".

Klart er
* $P(A) = 13/52 = 1/4$.
* $P(B) = 12/52 = 3/13$.

Den felles begivenheten $D = A\cap B$ = "mengden av alle ruter-bildekort", som tilfredsstiller både $A$ og $B$, inneholder åpenbart de følgende 3 kortene:
* "ruter knekt"
* "ruter dame"
* "ruter konge".

Altså er
$$P(D) = 3/52.$$


----------------------------------
__Spørsmål:__<br>
La den kombinerte begivenheten $C = A \cup B$ = "mengden av alle kort som enten er ruter eller bildekort".

Hva er $$P(C)=P(A \cup B)?$$

Merk at $P(A) + P(B) = 25/52$ ikke er det riktige svaret, siden vi teller sannsynlighetene for $D$ = "mengden av alle ruter-bildekort" to ganger.

Situasjonen er illustrert i [Venn-diagrammet](https://en.wikipedia.org/wiki/Venn_diagram) i figur 3.4 i OIS:

<img src="VennDiagram3_4.png"/>

Åpenbart inneholder begivenheten $A\cup B$ $13+9=22$ kort. Siden begivenheten $D = A\cap B$ bidrar to ganger – både i $P(A)$ og $P(B)$, er den riktige sannsynligheten for $C = A\cup B$
$$P(C) = P(A\cup B) = P(A) + P(B) - P(D)$$
$$ =P(A) + P(B) - P(A\cap B)= 13/52 + 12/52 - 3/52 = 22/52 = 11/26.$$
Dette resultatet er et spesialtilfelle av

__Den generelle addisjonsregelen for sannsynlighet__<br>
Hvis $A$ og $B$ er to vilkårlige begivenheter, disjunkte eller ikke, så er sannsynligheten for at minst en av dem vil inntreffe i et tilfeldig eksperiment
$$P(A\cup B) = P(A) + P(B) - P(A\cap B),$$
der $P(A \cap B)$ er sannsynligheten for at den felles begivenheten inntreffer i det tilfeldige eksperimentet.

#### Komplementet til en begivenhet
Å kaste en terning gir en verdi i mengden $S =\{1, 2, 3, 4, 5, 6\}$. Denne mengden av alle mulige utfall kalles __utfallsrommet__ ($S$) for å kaste en terning. Noen ganger er det nyttig å undersøke utfallsrommet for å identifisere komplementet til en begivenhet.

La $D = \{2, 3\}$ representere begivenheten at utfallet av et terningkast er 2 eller 3. Da representerer __komplementet__ $D^c$ til $D$ alle utfall i utfallsrommet $S$ som ikke er i $D$. Disse  betegnes med $$D^c = S\setminus D = \{1, 4, 5, 6\}.$$

<img src="Complement3_9.png"/>

Komplementet til en begivenhet $A$ har to svært viktige egenskaper:
1. Hvert mulige utfall som ikke er i $A$, er i komplementet $A^c$ (og omvendt), og
2. de to begivenhetene $A$ og $A^c$ er disjunkte.

Siden utfallsrommet $S = A\cup A^c$, gjelder følgende alltid for komplementer:

* $ 1 = P(S) = P(A\cup A^c) = P(A) + P(A^c)=1,$ og
* $P(A) = 1 - P(A^c).$

__Merknad:__<br>
I enkle eksempler er det gjennomførbart å beregne $A$ eller $A^c$ i noen få trinn. Imidlertid kan bruk av komplementet forenkle beregninger etter hvert som problemene vokser i kompleksitet.

#### Uavhengighet
Vi sier at to begivenheter [$A$ og $B$ er __uavhengige__](https://en.wikipedia.org/wiki/Independence_(probability_theory)#Two_events) når sannsynligheten for deres felles begivenhet $A\cap B$ tilfredsstiller multiplikasjon
$$P(A\cap B) = P(A)P(B).$$

__Merk__:<br>
To tilfeldige prosesser sies å være __uavhengige__ hvis det å kjenne utfallet av den ene ikke gir nyttig informasjon om utfallet av den andre. 

For eksempel er det å kaste en mynt og kaste en terning to uavhengige prosesser – å vite at mynten var kron hjelper ingenting med å avgjøre det faktiske utfallet av et terningkast. På den annen side beveger ulike aksjekurser seg ofte opp eller ned samtidig, så de betraktes ikke uavhengige.

Å kaste to terninger er også et eksempel på uavhengige prosesser. Med 
* $A_1=\{(1,1),\ (1,2),\ ..., \ (1,6)\}$ (begivenheten at den første terningen er $1$) 
* $B_1=\{(1,1),\ (2,1),\ ..., \ (6,1)\}$ (begivenheten at den andre terningen er $1$)

har vi $A_1\cap B_1 = \{(1,1)\}$ der

* $P(A_1) = \frac{6}{36} = \frac{1}{6}$
* $P(B_1) = \frac{6}{36} = \frac{1}{6}$, og
* $P(A_1\cap B_1) = \frac{1}{36} = P(A_1)P(B_1)$.

Med $k$ begivenheter $A_1,\ A_2,\ ...,\  A_k$ som tilsvarer $k$ uavhengige tilfeldige prosesser, er __sannsynligheten for at de inntreffer samtidig__ gitt ved produktet av enkelt-begivenhetene:
$$P(A_1\cap A_2\cap ... \cap A_k) = P(A_1)P(A_2)\cdots P(A_k).$$

__OIS-Eksempel 3.25__<br>
Spørsmål: Anta at vi tilfeldig trekker et kort fra en kortstokk. Er begivenheten $A=$"kortet er hjerter" uavhengig av begivenheten $B=$"kortet er et ess"? 

Åpenbart har vi $P(A)=\frac{13}{52}=\frac{1}{4}$, $P(B)=\frac{4}{52}=\frac{1}{13}$ og begivenheten $C = A\cap B=$"kortet er ess hjerter" der $P(A\cap B)=\frac{1}{52}$. Altså,
$$P(A)P(B) = \frac{1}{13}\cdot \frac{1}{4}=\frac{1}{52}=P(A\cap B).$$
Siden likheten holder, følger det at begivenhetene "kortet er hjerter" ($A$) og "kortet er et ess" ($B$) uavhengige.
