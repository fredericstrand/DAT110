## **1.1 Case-studie: Bruk av stenter for å forhindre hjerneslag**

Eksemplet som introduseres i denne delen er en studie publisert i [New England Journal of Medicine 15. september 2011](https://www.nih.gov/news-events/nih-research-matters/preventing-second-stroke) som undersøker __følgende forskningsspørsmål__:

## **1.3 Prinsipper og strategier for utvalg**

Statistisk baserte konklusjoner er avhengig av data samlet inn fra utvalg av populasjoner. Denne delen diskuterer hvordan utvalg gjennomføres for å sikre god representativitet i et statistisk utvalg.

### [**Populasjon vs. Utvalg**](https://en.wikipedia.org/wiki/Sampling_(statistics))
- [**Populasjon**](https://en.wikipedia.org/wiki/Sampling_(statistics)#Population_definition): Hele gruppen vi ønsker å studere.
- **Utvalg**: En delmengde av populasjonen som brukes til analyse.

<img src="SamplingFromPopulation.png" width="500" height="300"/>

----

### [**Utvalgsmetoder**](https://en.wikipedia.org/wiki/Sampling_(statistics))
1. [**Enkel tilfeldig utvelgelse**](https://en.wikipedia.org/wiki/Sampling_(statistics)#Simple_random_sampling): Hver enhet har lik sannsynlighet for å bli valgt.

<img src="Random.png" width="600" height="300"/>

2. [**Stratifisert utvalg**](https://en.wikipedia.org/wiki/Sampling_(statistics)#Stratified_sampling): Populasjonen deles inn i strata (delpopulasjoner) basert på egenskaper, og et tilfeldig utvalg tas fra hvert stratum.

<img src="Stratified.png" width="600" height="300"/>

3. [**Klyngeutvalg**](https://en.wikipedia.org/wiki/Sampling_(statistics)#Cluster_sampling): Populasjonen deles inn i klynger (clustre), og hele klynger velges tilfeldig.

<img src="Cluster.png" width="600" height="300"/>

4. [**Flertrinnsutvalg**](https://en.wikipedia.org/wiki/Multistage_sampling): Vi samler et tilfeldig utvalg innenfor hver tilfeldig valgte klynge.

<img src="Multistage.png" width="600" height="300"/>


### [**Utvalgsfeil**](https://en.wikipedia.org/wiki/Sampling_(statistics)#Errors_in_sample_surveys)
Utvalgsfeil oppstår når utvalget som gjøres ikke representerer populasjonen man ønsker å trekke konklusjoner om ikke er nøyaktig (tilstrekkelig dekkende). Dette kan både skyldes skjevheter i utvalgsprosessen og tilfeldigheter.

## [**1.4 Eksperimenter**](https://en.wikipedia.org/wiki/Experiment)

Eksperimenter brukes til å undersøke årsakssammenhenger mellom variabler.

### **Observasjonsstudier** vs. **Eksperimenter**
- [**Observasjonsstudier**](https://en.wikipedia.org/wiki/Observational_study): Data samles inn uten å påvirke hvordan de oppstår.
- [**Eksperimenter**:](https://en.wikipedia.org/wiki/Design_of_experiments) Forskere manipulerer forklaringsvariabler for å studere deres effekt på responsvariable man er interessert i.

### [**Randomiserte eksperimenter**](https://en.wikipedia.org/wiki/Design_of_experiments#Randomized_experiments)
I randomiserte eksperimenter blir deltakerne tilfeldig fordelt på grupper, noe som reduserer skjevhet og øker troverdigheten til årsaksslutninger.

### __De fire prinsippene for eksperimentell design__
* [__Kontroll:__](https://en.wikipedia.org/wiki/Controlling_for_a_variable#Experiments) Forskere tildeler behandlinger til tilfeller og forsøker å kontrollere for andre forskjeller mellom grupper. Dette kan innebære å standardisere forhold som kan påvirke resultatene, for eksempel mengden vann pasienter drikker når de tar en pille.
* [__Randomisering:__](https://en.wikipedia.org/wiki/Randomization) Deltakerne blir tilfeldig fordelt på behandlingsgrupper for å balansere forskjeller som ikke kan kontrolleres, for eksempel individuelle helsevaner. Randomisering bidrar til å forhindre skjevhet i studien.
* [__Replikering:__](https://en.wikipedia.org/wiki/Replication_(statistics)) Ved å observere flere tilfeller kan forskere mer nøyaktig estimere effekten av forklaringsvariabelen på responsen. Dette innebærer å samle inn et stort nok utvalg, og andre forskere kan også gjenta studien for å bekrefte funnene.
* [__Blokking:__](https://en.wikipedia.org/wiki/Blocking_(statistics)) Når forskere mistenker at andre variabler enn behandlingen påvirker responsen, kan de gruppere individer basert på disse variablene i blokker og deretter randomisere innenfor hver blokk til behandlingsgruppene. Dette sikrer at hver behandlingsgruppe har en balansert representasjon i forhold til disse variablene.

Se også [Fishers prinsipper](https://en.wikipedia.org/wiki/Design_of_experiments#Fisher's_principles).

----

#### [__Reduksjon av skjevhet__](https://en.wikipedia.org/wiki/Bias_(statistics)#Addressing_statistical_bias) __i menneskekontrollerte eksperimenter__
Randomiserte eksperimenter er grunnleggende for god datainnsamling, men de sikrer ikke et objektivt perspektiv på årsak- og virkningsforholdet i alle tilfeller.

Menneskekontrollerte studier er perfekte eksempler der [skjevhet](https://en.wikipedia.org/wiki/Bias_(statistics)) utilsiktet kan oppstå. Her ser vi på en studie der et nytt legemiddel ble brukt til å behandle hjerteinfarktpasienter. Spesielt ønsket forskerne å vite om legemidlet reduserte dødsfall hos pasienter.

De designet et randomisert eksperiment fordi de ønsket å trekke kausale konklusjoner om legemidlets effekt. Studiedeltakere ble tilfeldig plassert i to studiegrupper. En gruppe, __behandlingsgruppen__, fikk legemidlet. Den andre gruppen, kalt __kontrollgruppen__, fikk ingen legemiddelbehandling.

I en studie deles deltakerne inn i behandlings- og kontrollgrupper. Behandlingsgruppen får det faktiske legemidlet, mens kontrollgruppen ikke gjør det, noe som kan føre til emosjonelle effekter som kan påvirke studien.

__Blindtesting__<br>
For å forhindre denne skjevheten holdes deltakerne uvitende om gruppetildelingen sin, en praksis kjent som [blindtesting](https://en.wikipedia.org/wiki/Blinded_experiment).

Men hvis kontrollgruppen ikke får noen behandling, vil de kjenne til statusen sin. For å løse dette gis placebo – falske behandlinger som sukkerpiller – til kontrollgruppen for å opprettholde blindtesting.

Placebo kan forårsake reelle forbedringer, kjent som placeboeffekten. Videre må forskere og leger også blindes for å forhindre at de utilsiktet påvirker studien ved å behandle pasienter forskjellig basert på deres kunnskap om hvem som får den faktiske behandlingen. Dette oppnås gjennom et [dobbeltblindt](https://en.wikipedia.org/wiki/Blinded_experiment#Terminology) oppsett, der verken pasientene eller helsepersonellet vet hvem som får den virkelige behandlingen.

### Noen [vanlige sannsynlighetsfordelinger](https://en.wikipedia.org/wiki/Probability_distribution#Common_probability_distributions_and_their_applications)
Vi skal lære om sannsynligheter og sannsynlightesfordelinger i kapittel 3 og 4. Sannsynlighetsfordelingene vi skal jobbe med er enten "kontinuerlige" eller "diskrete". 

Uansett, hvis $f$ er en sannsynlighetsfordeling er $f(x)\geq 0$ for alle $x$ i sitt definisjonsoråde. 
* Dersom $f$ er en [diskret sannsynlighetsfordelig](https://en.wikipedia.org/wiki/Probability_mass_function), dvs er definert for $x\in \{x_1, x_2,...\}$, så er $$\sum_x f(x) =1.$$
* Dersom $f$ er en [kontiniuerlig sannsynlighetsfordelig](https://en.wikipedia.org/wiki/Probability_density_function), også kalt en sannsynlighets tetthetsfunksjon, dvs er definert for $x\in [a, b]$, så er $$\int_a^b f(x)dx =1.$$

----

1. [**Normal-fordelinger**](https://en.wikipedia.org/wiki/Normal_distribution):
   - Symmetrisk klokkeformet graf.
   - Beskrevet og bestemt ved gjennomsnitt ($\mu$) og standardavvik ($\sigma$).
   - Sannsynlighets tetthetsfunksjon:
     $$
     f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}
     $$
     hvor $-\infty<x<\infty$.

```julia
# Julia - kode:
μ = 0; σ = 1; 

f(x) = pdf.(Normal(μ, σ), x) # Sannsynligets tetthetsfordelingen til stadard normalfordelingen
F(x) = cdf.(Normal(μ, σ), x) # Den korresponderende kumulative fordelingsfunksjonen
q(x) = quantile.(Normal(μ, σ), x) # Kvantilfuksjonen til stadard normalfordelingen

#  Til numeriske integrasjon kan vi benytte QuadGK-pakka:
using QuadGK  # Load QuadGK package
qlow = q(0.025); #  2.5% kvantilen
qhigh = q(0.975); # 97.5% kvantilen

I = quadgk(f, qlow, qhigh) # Returnerer tuppel med: (integral-verdien, feil_estimat)

qlow, qhigh, I[1], F(qhigh)-F(qlow), q(0), q(1), quadgk(f, q(0),q(0.2))[1]
```

## Mål for sentraltendens og spredning

Dette skal vi lære om i kapittel 4.

Anta at et (statistisk uavhengig) utvalg av $n$ observasjoner ${\displaystyle x_{1},x_{2},\ldots ,x_{n}}$ er tatt fra en statistisk populasjon.

### [Mål for sentraltendens](https://en.wikipedia.org/wiki/Central_tendency#Measures)
1. [Gjennomsnitt](https://en.wikipedia.org/wiki/Arithmetic_mean) ($\bar{x}$):
   $$
   \bar{x} = \frac{\sum x_i}{n}
   $$
2. [Median](https://en.wikipedia.org/wiki/Median): Midterste verdi når de $n$ observerte data-verdiene er sortert.
3. [Modus](https://en.wikipedia.org/wiki/Mode_(statistics)): Den hyppigste verdien (verdien med flest forekomster) blant ${\displaystyle x_{1},x_{2},\ldots ,x_{n}}$.

### [Mål for statistisk spredning](https://en.wikipedia.org/wiki/Statistical_dispersion)
1. [Utvalgsvarians](https://en.wikipedia.org/wiki/Variance#Sample_variance) ($s^2$):
   $$
   s^2 = \frac{\sum (x_i - \bar{x})^2}{n}
   $$
2. [Standardavvik](https://en.wikipedia.org/wiki/Standard_deviation) ($s$):
   $$
   s = \sqrt{s^2}
   $$

### [Standardfeilen til et utvalgsgjennomsnitt $\bar{x}$](https://en.wikipedia.org/wiki/Standard_error) ($SE$):
   $$
   SE = \frac{s}{\sqrt{n}}
   $$

Dette danner et nyttig utgangspunkt for å jobbe videre med statistikk. Vi vil seinere introdusere grunnleggende begreper rundt datainnsamling, organisering og analyse, samt understreke viktigheten av riktig utvalg og eksperimentell design for å trekke gyldige konklusjoner fra data.

```julia
using Statistics, StatsBase  # For statistiske funksjoner

# ----------------------------------------------------------------

# Generer tilfeldige tall
n = 30  # Antall tall
lower_bound = 0  # Nedre grense for tilfeldige tall
upper_bound = 10 # Øvre grense for tilfeldige tall

# Genererer n tilfeldige heltall mellom lower_bound og upper_bound
random_numbers = rand(lower_bound:upper_bound, n)  
println("Tilfeldige tall: ", random_numbers)

# -----------------------------------------------------------------

# Beregn gjennomsnitt
mean_value = mean(random_numbers)
println("\nGjennomsnitt: ", mean_value)

# Beregn median
median_value = median(random_numbers)
println("Median: ", median_value)

# Beregn modus (krever litt mer arbeid, siden Julia ikke har en standard "mode"-funksjon)
# using StatsBase  # For å bruke 'mode' funksjonen
mode_value = mode(random_numbers)
println("Modus: ", mode_value)

# Beregn standardavvik
std_dev = std(random_numbers)
println("Standardavvik: ", std_dev)

# Beregn varians
variance = var(random_numbers)
println("Varians: ", variance)

# Beregn standardfeilen til gjennomsnittet:
sem = std_dev/sqrt(n)
println("Standardfeilen til gjennomsnittet: ", sem)
```

## __Statistisk hypotesetesting__
[Statistisk hypotesetesting](https://en.wikipedia.org/wiki/Statistical_hypothesis_test) (tema i kapittel 5-7) representerer et systematisk rammeverk for å evaluere hypoteser ved hjelp av data og sannsynlighetsvurderinger. Dette er en metodikk som gir grunnlag for å ta beslutninger eller trekke konklusjoner om en populasjon basert på et utvalg data. De grunnleggende ideene og prinsippene er som følger:  

---

### **1. Nullhypotese ($H_0$) og alternativ hypotese ($H_1$):**  
- **[Nullhypotese ($H_0$)](https://en.wikipedia.org/wiki/Null_hypothesis)**: Representerer en påstand eller en standard antakelse som man ønsker å teste. Typisk innebærer dette ingen effekt eller ingen forskjell (f.eks. "ingen sammenheng mellom variabler" - eller "ingen forskjell mellom grupper") basert på målte data.  
- **[Alternativ hypotese ($H_1$)](https://en.wikipedia.org/wiki/Alternative_hypothesis)**: Representerer et alternativ til nullhypotesen, ofte noe man ønsker å bevise (f.eks. "det er en sammenheng mellom variabler -  eller "det er en forskjell mellom grupper") basert på målte data.  

---

### **2. Testobservator:**  
En [__testobservator__](https://en.wikipedia.org/wiki/Test_statistic) er en matematisk størrelse (en verdi) som __beregnes fra måledata__, og som under __rimelige antagelser om usikkerhet__ vurderes i forhold til en __sannsynlighetsfordeling__ som er __forenlig med at nullhypotesen er korrekt__. Testobservatorene vi jobber med i DAT110 vurderes gjerne mot følgende fordelinger:  
- [Normalfordelinger](https://en.wikipedia.org/wiki/Normal_distribution) ($Z$-observatorer, når utvalgene er relativt store)  
- [$T$-fordelinger](https://en.wikipedia.org/wiki/Student%27s_t-distribution)   ($t$-observatorer, mest brukt for små utvalg)  
- [$χ²$-fordelinger](https://en.wikipedia.org/wiki/Chi-squared_distribution)  ($χ²$-observatorer, typisk for bahandling av kategoriske data)  

---

### **3. Signifikansnivå ($\alpha$):**  
- Et [signifikansnivå](https://en.wikipedia.org/wiki/Statistical_significance) er en forhåndsdefinert toleranseverdi (notasjon: $\alpha$) på __sannsynligheten for feilaktig å forkaste nullhypotesen__ når den faktisk er sann ([__Type I-feil__](https://en.wikipedia.org/wiki/Type_I_and_type_II_errors)).  
- Typiske verdier for $\alpha$ er små tall som 0.05 (5 %) eller 0.01 (1 %) siden man ønsker å gjøre denne typen feil sjelden.  

---

### **4. P-verdi:**  
- En [p-verdi](https://en.wikipedia.org/wiki/P-value) er en beregnet sannsynlighet for at en teoretisk testobservator er minst så ekstrem som den observerte verdien beregnet fra data forutsatt at nullhypotesen er sann.  
- Hvis **p-verdi < $\alpha$**, forkaster man nullhypotesen og konkluderer med at resultatet er statistisk signifikant.  
- Hvis **p-verdi ≥ $\alpha$**, forkastes ikke nullhypotesen.  

---

### **5. [Testresultat og beslutning](https://en.wikipedia.org/wiki/Statistical_hypothesis_test#Performing_a_frequentist_hypothesis_test_in_practice):**  
- **Forkasting av $H_0$:** Det er tilstrekkelig bevis til å støtte $H_1$.  
- **Ikke forkasting av $H_0$:** Det er ikke nok bevis til å støtte $H_1$. Dette betyr ikke nødvendigvis at nullhypotesen er sann, men at dataene ikke gir tilstrekkelig beslutningsgrunnlag for å forkaste den.  

---

### **Eksempel på hypotesetesting:**  
Anta at man vil teste om en __ny medisin__ har en __effekt på blodtrykk__.  
- **$H_0$:** Den nye medisinen har ingen effekt på blodtrykk.  
- **$H_1$:** Den nye medisinen har en effekt på blodtrykk.  
- Vi bestemmer det ønskede signifikansnivået ($\alpha$) for testen.

Vi samler deretter inn data, beregner verdien til vår testobservator og finner den aktuelle p-verdien til testobservatoren.  
- Hvis den observerte p-verdien er 
    - lavere enn signifikansnivået $\alpha$ ($0.05$) - vi forkaster nullhypotesen og konkluderer med at medisinen har en effekt. 
    - høyere enn signifikansnivået $\alpha$ ($0.05$) - vi beholder nullhypotesen og konkluderer med at innsamlede data ikke gir grunnlag for å påstå at medisinen har effekt.

Under $H_0$ forventer vi at testobservatoren vår oppfører seg i henhold til en kjent sannsynlighetsfordeling (nevnt over), og forkaster $H_0$ hvis verdien til testobservatoren vår kan betraktes som lite sannsynlig (p-verdien $<\alpha$).

---

```julia
#using Pkg
#Pkg.add("Plots")

using Plots
using Distributions

μ = 0;
σ = 1;

f(x) = pdf.(Normal(μ, σ), x) # Sannsynligets tetthetsfordelingen til stadard normalfordelingen
q(x) = quantile.(Normal(μ, σ), x) # Kvantilfuksjonen til stadard normalfordelingen

x = -4:0.01:4    # x-verdier for plotting.
α = 0.05;        # Kritisk grense (α = 0.05, tosidig)
q(α/2), q(1-α/2) # nedre og øvre α/2-kvantiler i standard normalfordelingen

# Plot av sannsynlighets tetthets funksjonen (pdf)
plot(x, f.(x), label="Normalfordeling", linewidth=2, xticks=false, yticks=false, bottom_margin=5Plots.mm, left_margin=5Plots.mm)

# Markering av områder med lite sannsylige verdier for testobservatoren 
plot!(x[x .<= q(α/2)], f.(x[x .<= q(α/2)]), fillrange = 0, fillalpha=0.3, label="Forkastningsområde", color=:red)
plot!(x[x .>= q(1-α/2)], f.(x[x .>= q(1-α/2)]), fillrange = 0, fillalpha=0.3, label="", color=:red)
vline!([q(α/2), q(1-α/2)], linestyle=:dash, color=:black, label="")

#xlabel!("Z-verdi")
#ylabel!("Tetthet (pdf)")
title!("Hypotesetesting: Testobservator-verdier og forkastning")
```

### **Standardfeil og presisjon:**

Utvalgsstørrelsen (mengden innsamlede data) påvirker presisjon og kvaliteten til en hypotesetest. 

**Standardfeilen** er et mål på hvor mye utvalgsobservatoren (f.eks. gjennomsnittsverdien til de innsamlede dataene) forventes å variere hvis man hadde gjentatt datainnsamlingen mange ganger.  

La oss anta at $\sigma$ er standardavviket i populasjonen vi samler data fra.  

Når utvalgsstørrelsen ($n$) øker, reduseres standardfeilen til utvalgsobservatoren (gjennomsnittsverdien) som følger:  
$$
SE = \frac{\sigma}{\sqrt{n}}
$$

**Observasjon:** Større utvalg gir mindre spredning og mer presise estimater av populasjonsparameterne.  

```julia
x = -5:0.01:5

σ = 8           # σ² er populasjonsvariansen

nₗ = 30         # Et lite utvalg
SEₗ = σ/sqrt(nₗ) # Standardfeilen til utvalgsobservatoren (empirisk gjennomsnitt)

nₛ = 150        # Et stort utvalg
SEₛ = σ/sqrt(nₛ)

fₗ(x) = pdf.(Normal(0, SEₗ), x)  # Større varians (lite utvalg)
fₛ(x) = pdf.(Normal(0, SEₛ), x)  # Mindre varians (stort utvalg)
q(x) = quantile.(Normal(0, 1), x)

plot(x, fₗ(x), xticks=true, yticks=false, label="Lite utvalg (n = $nₗ)", linewidth=2)
plot!(x, fₛ(x), label="Stort utvalg (n = $nₛ)", linewidth=2)
plot!(x[x .<= q(α/2)*SEₗ], fₗ.(x[x .<= q(α/2)*SEₗ]), fillrange = 0, fillalpha=0.3, label="Forkastningsområde", color=:blue)
plot!(x[x .>= q(1-α/2)*SEₗ], fₗ.(x[x .>= q(1-α/2)*SEₗ]), fillrange = 0, fillalpha=0.3, label="", color=:blue)
vline!([q(α/2)*SEₗ, q(1-α/2)*SEₗ], linestyle=:dash, color=:black, label="")

plot!(x[x .<= q(α/2)*SEₛ], fₛ.(x[x .<= q(α/2)*SEₛ]), fillrange = 0, fillalpha=0.3, label="Forkastningsområde", color=:red)
plot!(x[x .>= q(1-α/2)*SEₛ], fₛ.(x[x .>= q(1-α/2)*SEₛ]), fillrange = 0, fillalpha=0.3, label="", color=:red)
vline!([q(α/2)*SEₛ, q(1-α/2)*SEₛ], linestyle=:dash, color=:black, label="")

#xlabel!("Verdi")
#ylabel!("Tetthet")
title!("Effekt av utvalgsstørrelse på Forkastningsområde")
```


### **6. [Feiltyper](https://en.wikipedia.org/wiki/Type_I_and_type_II_errors) og [teststyrke (power)](https://en.wikipedia.org/wiki/Power_(statistics)#Description) ved hypotesetesting:**  
**[Type I-feil](https://en.wikipedia.org/wiki/Type_I_and_type_II_errors#Type_I_error):** Forkaste en sann nullhypotese (falsk positiv). <br> 
**[Type II-feil](https://en.wikipedia.org/wiki/Type_I_and_type_II_errors#Type_II_error):** Unnlate å forkaste en gal nullhypotese (falsk negativ).  

Det er 4 mulige scenarios ifbm statistisk hypotesetesting:
1.   Forkast  $H_0$ når den er sann (Type I-feil 👎).
2.   Ikke forkast $H_0$ når den er sann (rett 👍).
3.   Forkast $H_0$ når den er falsk (rett 👍).
4.   Ikke forkast $H_0$ når den er falsk (Type II-feil 👎).

|                      | Sannsynlighet for å forkaste  $H_0$ | Sannsynlighet for å ikke forkaste $H_0$ |
|----------------------|-------------------------------------|-----------------------------------------|
| ...når $H_0$ er sann   | $\alpha$ (Type I-feilsannsynlighet) 👎 | $1 - \alpha$   👍                     |
| ...når $H_1$ er sann   | $1 - \beta$ (Teststyrken)  👍               | $\beta$  (Type II-feilsannsynlighet)  👎  |

__Teststyrken__ ($1 - \beta$) er sannsynligheten til området der $H_1$ er riktig, og testen oppdager det.

---

```julia
σ = 15         # σ² er populasjonsvariansen
n = 50         # Forslag til utvalgsstørrelse
SE = σ/sqrt(n) # Standardfeilen til utvalgsobservatoren (empirisk gjennomsnitt)

# ---------------- Hypotesetesting ----------------------------
μ₀ = 0         # Forventningsverdi under H₀
μ₁ = 4         # Forventningsverdi under H₁
# -------------------------------------------------------------

x = Int(floor(μ₀-4*SE)):0.01: Int(ceil(μ₁+4*SE)) # x-verdier av interesse for plotting.

f₀(x) = pdf.(Normal(μ₀, SE), x) # tetthetsfordelingen til utvalgsobservatoren under H₀
f₁(x) = pdf.(Normal(μ₁, SE), x) # tetthetsfordelingen til utvalgsobservatoren under H₁     

F₀(x) = cdf.(Normal(μ₀, SE), x) # Kumulativ fordeling til utvalgsobservatoren under H₀ 
F₁(x) = cdf.(Normal(μ₁, SE), x) # Kumulativ fordeling til utvalgsobservatoren under H₁  

q(x) = quantile.(Normal(0, 1), x) # Kvantilfunksjonen til standard normalfordelingen.

# Kritiske grenser
α = 0.05 # type  I-feil krav
β = 0.2  # type II-feil krav
fill_xᵢ  = x[x .≥ q(1-α)*SE+μ₀]  # Kritisk område (Type I-feil) 
fill_xᵢᵢ = x[x .≤ q(β)*SE+μ₁]    # Kritisk område (Type II-feil) 

powₙ =  1-F₁(q(1-α)*SE+μ₀)
powₙ = round(powₙ, digits = 4)

# Plot begge fordelingene med markering av områdene for type I og type II feil.
plot(x,  f₀.(x), label="H₀", linewidth=2)
plot!(x, f₁.(x), label="H₁", linewidth=2)
plot!(fill_xᵢ,  f₀.(fill_xᵢ),   fillrange = 0, fillalpha=0.3, label="Type I-feil (α = $α)",  color=:blue)
plot!(fill_xᵢᵢ, f₁.(fill_xᵢᵢ),  fillrange = 0, fillalpha=0.3, label="Ønsket Type II-feil (β = $β)", color=:red)
plot!(fill_xᵢ, f₁.(fill_xᵢ),  fillrange = 0, fillalpha=0.3, label="Faktisk styrke = $powₙ", color=:green )
xlabel!("Testobservator")
ylabel!("Tetthet")
title!("Type I- og Type II-feil")

display(plot!())
```

__Spørsmål:__<br>
Er det mulig å finne en utvalgsstørrelse $n$ slik at testen tilfredsstiller kravene om at både 
 - Type I-feilen er mindre enn $\alpha$, og
 - Type II-feilen er mindre enn $\beta$ ?

 Svaret er JA!

 For problemet over kan løsningen bestemmes ved at de aktuelle kvantilverdiene i de to fordelingene settes lik hverandre
 $$q(1-α)SE+μ₀ = q(β)SE+μ₁,$$ 
 og likningen løses mhp utvalgsstørrelsen $n$ når vi substituerer $\frac{\sigma}{\sqrt{n}}$ for $SE$. 
 
 Ved å utnytte likheten $q(α) = -q(1-α)$ blir løsningen:

 $$n = \left(\frac{q(α)+q(β)}{μ₀ - μ₁}\right)^2σ^2.$$

```julia
# Løs (med hensyn på utvalgsstørrelsen (n) likningen 
# q(1-α)*SE+μ₀ = q(β)*SE+μ₁
# som setter (1-α)-kvantilen til f₀-fordelingen lik β-kvantilen til f₁-fordelingen.
# Dette sikrer at kravene til både type I-feil og type II-feil er oppfylt.

# Løsningen er at n må være lik minste heltall rundet opp fra nₜ: 
nₜ = ((q(α)+q(β))/(μ₀ - μ₁))^2*σ^2
nₜ = Int(ceil(nₜ))    # Runder løsningen opp til nærmeste heltall - som gir minste heltallige utvalgsstørrelse som er brukbar.

println("Løsningen er at utvalgsstørrelsen n må være minst n = $(ceil(nₜ)) for å oppnå ønsket teststyrke på $(1-β).")
println("Med n = $n oppnås en faktisk styrke på $powₙ.")
```

__Nummerisk nullpunktsløsning__

Problemet med å bestemme utvalgsstørrelse kan også løses ved å finne nullpunktet til funksjonen
$$N(n) = μ₀ + q(1-α)*σ/\sqrt{n} - μ₁ - q(β)*σ/\sqrt{n}.$$

Dette kan gjøres nummerisk ved hjelp av [Roots](https://juliamath.github.io/Roots.jl/stable/)-pakken i Julia:

```julia
# Løsningen nₜ er også nullpunktet til funksjonen N(n):
N(n) = μ₀ + q(1-α)*σ/sqrt(n) - μ₁ - q(β)*σ/sqrt(n)

#using Pkg
#Pkg.add("Roots")

# Bestemmer nullpunktet til N(n) nummerisk:
using Roots
nn  = 150              # Startverdi for søk etter den nummeriske løsningen n₀:
n₀ = find_zero(N, nn)  # Funksjonen returnerer en numerisk løsning til likningen N(n) = 0.
n₀ = Int(ceil(n₀))    # Runder løsningen opp til nærmeste heltall - som gir minste heltallige utvalgsstørrelse som er brukbar.

println("Løsningen er at utvalgsstørrelsen n må være minst n = $(ceil(n₀)) for å oppnå ønsket teststyrke på $(1-β).")
println("Med n = $n oppnås en faktisk styrke på $powₙ.")

# Plotter N(n) med markering av nullpunktet:
m = n₀-50:n₀+50
plot(m, N.(m), label ="", xlabel = "n - utvalgsstørrelse", ylabel = "N(n)")
scatter!([n₀], [0], label ="n₀ = $n₀")
title!("Numerisk løsning av likningen N(n) = 0")
display(plot!())

```
