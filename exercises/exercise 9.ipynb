## Task 1

```python
from scipy.stats import t

def calculate_critical_t(n,CL):
    df = n-1
    
    critical_t = t.ppf(1-(1-CL)/2, df)
    return critical_t

A = calculate_critical_t(6,0.9)
B = calculate_critical_t(21,0.98)
C =  calculate_critical_t(29,0.95)
D = calculate_critical_t(12,0.99)

print(f'A: {A}, B: {B}, C: {C}, D: {D}')
```

## Task 2

```python
def find_p(n, T):
    df = n - 1
    
    return 2 * t.sf(abs(T), df)

A = find_p(11,1.91)
B = find_p(17,-3.45)
C = find_p(7, 0.83)
D = find_p(28, 2.13)

print(f'A: {A}, B: {B}, C: {C}, D: {D}')
```
